```python
import pandas as pd
df = pd.read_csv("titanic.csv")
df.head()
semilla=42
```

```python
df.info()
```

```python
from ast import copy_location
colNum = ["Age", "SibSp", "Parch", "Fare"]
colCat = ["Pclass", "Sex", "Embarked"]
df[colNum + colCat]
```

```python
from sklearn.dummy import DummyClassifier
clasificador1 = DummyClassifier(strategy="most_frequent")
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

pn = make_pipeline(SimpleImputer(strategy="median"), StandardScaler())
pc = make_pipeline(SimpleImputer(strategy="most_frequent"), OneHotEncoder(handle_unknown="ignore",sparse_output=False))


prep = ColumnTransformer([
    ("num", pn, colNum),
    ("cat", pc, colCat)
])
```

```python
modelo = make_pipeline(prep, clasificador1)
```

```python
X = df[colNum + colCat]
y = df["Survived"]

X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=semilla)
```

```python
modelo.fit(X_train, y_train)
```

```python
y_pred = modelo.predict(X_test)
y_pred
```

```python
clasificador1.fit(X_train, y_train)
clasificador1.score(X_test, y_test)
acu = modelo.score(X_test, y_test)
acu
```

```python
from sklearn.svm import SVC
clasificador2=SVC()
modelo2=make_pipeline(prep, clasificador2)
modelo2.fit(X_train, y_train)
acu2=modelo2.score(X_test, y_test)
acu2
```
